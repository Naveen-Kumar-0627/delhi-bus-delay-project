```python
%sql
 
create catalog if not exists bus;
create database if not exists bus.data;
create volume if not exists bus.data.files;
```

```python
import os
li=["agency","calendar","routes","stop_times","stops","trips"]
path="/Volumes/bus/data/files/delhi-bus/"
for i in li:
    obsulte_path=f"{path}{i}.csv"
    if os.path.exists(obsulte_path):
        df=spark.read.format('csv')\
            .option("header",True)\
                .load(obsulte_path)
        df.write.format('delta')\
            .saveAsTable(f"bus.data.bronze_{i}")    
    else:
        print("file not found",i)
                


```
