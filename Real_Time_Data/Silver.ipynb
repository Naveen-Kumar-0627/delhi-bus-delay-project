```python
from pyspark.sql.types import *
from pyspark.sql import functions as F
```

```python
check="bus.data.checkp2"
 
```

```python
schema=StructType([
    StructField("vehicle_id",StringType()),
    StructField("trip_id",StringType()), 
    StructField("start_time",StringType()),
    StructField("state_date",StringType()),
    StructField("schedule_relationship",IntegerType()),
    StructField("route_id",StringType()),
    StructField("latitude",DoubleType()),
    StructField("longitude",DoubleType()),
    StructField("speed",DoubleType()),
    StructField("pulled_at",LongType()),
    StructField("ingest_at",LongType()),
    StructField("event_ts",LongType())
])
```

```python
def get_vehicle_data():
    df=spark.readStream.table("bus.data.bronze_bus_positions")
    df=df.withColumn("j",F.from_json(F.col("raw_json"),schema)).select(F.col("j.*"),F.col("j.state_date").alias("start_date")).drop("j.state_date")
    df.writeStream\
        .option("checkpointLocation",check)\
            .toTable("bus.data.silver_vehicle_data")
```
